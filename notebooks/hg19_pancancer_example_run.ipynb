# Window-based approach

```python
# List covariates to be used in the model
param_covariates = ['scaled_log_ncncm_rate']+\
                   ['scaled_log_perc_gc','scaled_log_perc_cpg', 'scaled_log_perc_tpc', 'scaled_log_perc_tpa']+\
                   ['scaled_log_perc_open']+\
                   ['scaled_yeo_rep']+['scaled_yeo_phylop'] +['scaled_log_expr']
```

```python
# Load the NCCM scan data and covariates
scan_df = nccm_enrichment.load_data('/Users/rappe860/Documents/work/NCCM-significance/hg19v3/3.negbin_regression/runs/w2kbp_phylop1.2/pancancer_ALL_whitelisted_nohyper_excl_melanoma_blood_n2101.2kbp_windows.phylop-1.2.scan.tsv')
covariates = nccm_enrichment.load_data('../data/hg19/hg19.2kbp_windows.covariates.tsv')#, w2k_expr_path]])

# Merge dfs and filter excluded genes
exclude = nccm_enrichment.load_data('../data/hg19/hg19.2kbp_windows.excluded_genes.txt')['gene'].tolist()
total_df = nccm_enrichment.merge_dfs_on_gene(scan_df, covariates)
total_df_filt = total_df[~total_df['gene'].isin(exclude)]
```

```python
# Define covariates to include in the model
covariates_list = param_covariates

# Update the parameters for the approach
params = nccm_enrichment.ModelParams(covariates_list=covariates_list, n_bins=10)

# Run the window-based approach
results_windows = nccm_enrichment.window_based_approach(total_df_filt, params)
```

```python
results_windows[results_windows['q_value']<0.1][['gene','closest_gene','q_value']]
```

# Gene-centric approach

```python
# List covariates to be used in the model
param_covariates = ['scaled_log_ncncm_rate']+\
                   ['scaled_log_perc_gc','scaled_log_perc_cpg', 'scaled_log_perc_tpc', 'scaled_log_perc_tpa']+\
                   ['scaled_log_perc_open']+\
                   ['scaled_yeo_rep']+['scaled_yeo_phylop'] +['scaled_log_expr']
```

```python
# Load the NCCM scan data and covariates for different flank sizes
scan_df_100 = nccm_enrichment.load_data('/Users/rappe860/Documents/work/NCCM-significance/hg19v3/3.negbin_regression/runs/f100kbp_phylop1.2/pancancer_ALL_whitelisted_nohyper_excl_melanoma_blood_n2101.phylop-1.2.scan.tsv')
scan_df_50 = nccm_enrichment.load_data('/Users/rappe860/Documents/work/NCCM-significance/hg19v3/3.negbin_regression/runs/f50kbp_phylop1.2/pancancer_ALL_whitelisted_nohyper_excl_melanoma_blood_n2101.50kbp_flanks.phylop-1.2.scan.tsv')
scan_df_10 = nccm_enrichment.load_data('/Users/rappe860/Documents/work/NCCM-significance/hg19v3/3.negbin_regression/runs/f10kbp_phylop1.2/pancancer_ALL_whitelisted_nohyper_excl_melanoma_blood_n2101.10kbp_flanks.phylop-1.2.scan.tsv')
scan_df_5 = nccm_enrichment.load_data('/Users/rappe860/Documents/work/NCCM-significance/hg19v3/3.negbin_regression/runs/f5kbp_phylop1.2/pancancer_ALL_whitelisted_nohyper_excl_melanoma_blood_n2101.5kbp_flanks.phylop-1.2.scan.tsv')

exclude_100 = nccm_enrichment.load_data('../data/hg19/hg19_gene_100kb_flanks.excluded_genes.txt')['gene'].tolist()
exclude_50 = nccm_enrichment.load_data('../data/hg19/hg19_gene_50kb_flanks.excluded_genes.txt')['gene'].tolist()
exclude_10 = nccm_enrichment.load_data('../data/hg19/hg19_gene_10kb_flanks.excluded_genes.txt')['gene'].tolist()
exclude_5 = nccm_enrichment.load_data('../data/hg19/hg19_gene_5kb_flanks.excluded_genes.txt')['gene'].tolist()

covariates_100 = nccm_enrichment.load_data('../data/hg19/hg19.100kbp_flanks.covariates.tsv')#, w2k_expr_path]])
covariates_50 = nccm_enrichment.load_data('../data/hg19/hg19.50kbp_flanks.covariates.tsv')#, w2k_expr_path]])
covariates_10 = nccm_enrichment.load_data('../data/hg19/hg19.10kbp_flanks.covariates.tsv')#, w2k_expr_path]])
covariates_5 = nccm_enrichment.load_data('../data/hg19/hg19.5kbp_flanks.covariates.tsv')#, w2k_expr_path]])

# Merge and filter the dfs
total_df_100 = nccm_enrichment.merge_dfs_on_gene(scan_df_100, covariates_100)
total_df_filt_100 = total_df_100[~total_df_100['gene'].isin(exclude_100)]

total_df_50 = nccm_enrichment.merge_dfs_on_gene(scan_df_50, covariates_50)
total_df_filt_50 = total_df_50[~total_df_50['gene'].isin(exclude_50)]

total_df_10 = nccm_enrichment.merge_dfs_on_gene(scan_df_10, covariates_10)
total_df_filt_10 = total_df_10[~total_df_10['gene'].isin(exclude_10)]

total_df_5 = nccm_enrichment.merge_dfs_on_gene(scan_df_5, covariates_5)
total_df_filt_5 = total_df_5[~total_df_5['gene'].isin(exclude_5)]

# Define covariates to include in the model
covariates_list = param_covariates

# Update the parameters for the approach
params = nccm_enrichment.ModelParams(covariates_list=covariates_list, n_bins=10)

# Run the gene-centric approach
results_flanks = nccm_enrichment.gene_centric_approach(total_df_filt_100,total_df_filt_50, total_df_filt_10, total_df_filt_5, params=params)
```
